```python
# load olist_geolocation_dataset file
df = pd.read_csv('/content/drive/MyDrive/olist-ecommerce-analytics/processed_csv/olist_geolocation_dataset.csv')
```

```python
# check data types
df.dtypes
```

```python
# validate Brazilian geographic coordinates
df['geolocation_lat'].between(-34, 6).all()
df['geolocation_lng'].between(-74, -34).all()
```

```python
# remove records with invalid coordinates
df = df[
    df['geolocation_zip_code_prefix'].between(0, 99999) &
    df['geolocation_lat'].between(-34, 6) &
    df['geolocation_lng'].between(-74, -34)
]
```

```python
# validate Brazilian state codes
df['geolocation_state'].unique()

valid_states = {
    'AC','AL','AP','AM','BA','CE','DF','ES','GO','MA','MT','MS',
    'MG','PA','PB','PR','PE','PI','RJ','RN','RS','RO','RR','SC',
    'SP','SE','TO'
}

invalid_states = df.loc[~df['geolocation_state'].isin(valid_states), 'geolocation_state'].unique()
print(invalid_states)
```

```python
# validate ZIP code prefix values
df['geolocation_zip_code_prefix'].between(0, 99999).all()
```
